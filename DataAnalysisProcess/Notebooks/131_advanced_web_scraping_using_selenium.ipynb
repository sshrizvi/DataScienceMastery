### 🎯 **Selenium**

***Selenium*** is an open-source tool primarily designed for *automating web browsers*. Unlike ***BeautifulSoup*** (which parses HTML), Selenium controls a real browser, executes JavaScript, interacts with dynamic web elements.  

Selenium was developed in **2004** by *Jason Huggins* at *ThoughtWorks*. It was originally created for web testing automation.  

Selenium is widely used by:  
1. QA Engineers, for testing.
2. Data Engineers / Data Scientists, for scraping.
3. Automation Engineers.

### 🎯 **Core Pillars of Selenium**

Following are the core pillars of Selenium that are significant in context of *Web Scraping*:  
1. **WebDriver (The Brain)**  
   The core engine that controls the browser. Selenium provides drivers for different browsers on their website. For example, ChromeDriver is used for Chrome, and GeckoDriver is used for Firefox.

2. **Browser (Execution Environment)**  
   A browser is needed for Selenium to execute or automate things. Selenium actually opens a browser. Selenium offers two modes of opening browsers:
   - Headed (Browser is visible)
   - Headless (Browser is not visible)

3. **DOM (Document Object Model)**  
   DOM is the structure of the webpage. Everything that we will scrape comes from HTML elements, attributes, hierarchy. Without DOM, Selenium is useless.

4. **Locators (Finding Elements)**  
   The MOST important skill in Selenium. Locators are used to locate elements in the DOM. Various locators that can be used for the purpose are as follows:
   - ID
   - Class Name
   - XPath
   - CSS Selector

5. **Waits**  
   Dynamic websites load content asynchronously. Waits are used as a preventive measure for *Element not found* errors.

6. **Actions**  
   Selenium doesn't just reads. It interacts. Various actions that Selenium can do are as follows: 
   - Click Buttons
   - Fill Forms
   - Scroll Pages
   - Hover
   - Press Keys

7. **JavaScript Execution**  
   The most advanced scraping power of Selenium is that it can executes JavaScript in the browser, that is being controlled by Selenium. This feature is used for:
   - Infinite Scroll
   - Triggering Hidden Elements
