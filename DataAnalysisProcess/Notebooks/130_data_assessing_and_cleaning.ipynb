### 🎯 **Unclean Data**

There are majorly two types of unclean data that we can think of:  
1. **Dirty Data**  
   Dirty Data, is also known as *Low Quality Data* (or Data with Quality Issues). This kind of data has content issues like:
      1. Duplicated Data
      2. Inaccurate Data
      3. Missing Data
      4. Corrupt Data  
   
   For Example, A table contains the data of physical properties (height, weight, color) of some students.  
   Some of the values in the `height` column, are negative values, which is irrational (or inaccurate).

2. **Messy Data**  
   Messy Data, also known as *Untidy Data* (or Data with Tidiness Issues). This kind of data has structural issues. Tidy Data has the following properties:
      1. Each variable forms a column.
      2. Each observation forms a row.
      3. Each obervational unit forms a table.

```python
[1, female, Zoe, Wellish, 576 Brown Bear Drive, Rancho California, California, 92390.0, United States, 951-719-9170ZoeWellish@superrito.com, 7/10/1976, 121.7 ,66 ,19.6]
```

#### **4. describe Method**

The `describe` method of Pandas provides us with the statistical summary of numerical columns in the DataFrame.  
With this, we can find issues in the distribution of columns.

````python
patients_df.describe()
````

````python
suspect = patients_df[patients_df['weight'] == patients_df['weight'].min()].copy()
suspect['calculated_bmi'] = suspect['weight'] / (suspect['height'] ** 2) * 703
suspect[['given_name', 'surname', 'weight', 'height', 'bmi', 'calculated_bmi']]
````

**Observations**  
1. Column `zip_code` contains a minimum value of 4-digits, which is wrong.
2. Column `weight` contains a minimum value of **48.8 Pounds**, equivalent to **22.13531 Kg**, which is suspicious.  
   On looking for that person details about `weight` and `height`, it is found that the weight information is incorrect.  
   This is verified by calculating the `bmi` of that person on the basis of given weight & height, which mismatches with the `bmi` in the data.  
   **Note :** This raises a concern that there is possibility that most of the `bmi` are calculated wrong.

So, now lets check for the `bmi` calculations issue.

````python
def calc_bmi(df):
    '''Calculate BMI.'''
        
    return round(df['weight'] / (df['height'] ** 2) * 703, 1)
````

````python
df = patients_df.copy()
df['calculated_bmi'] = calc_bmi(df)
incorrect_bmi_mask = df['bmi'] != df['calculated_bmi']
df[incorrect_bmi_mask]
````

**Observation :** There are two person (patient_id = 5, 211) with unusal weight & height values. So, lets document it.

Now, lets analyze the `treatments_df` using `describe` method.

````python
treatments_df.describe()
````

Since, `hba1c_change` is calculated as $hba1c_{start} - hba1c_{end}$. We can verify if there are any mistakes in calculation of `hba1c_change`.

````python
df = treatments_df.copy()
df['hba1c_calculated'] = (df['hba1c_start'] - df['hba1c_end'])
wrong_hba1c_change = (
    df['hba1c_change'].notnull() &
    ~np.isclose(a=df['hba1c_change'], b=df['hba1c_calculated'],
                rtol=1e-5, atol=1e-8)
)
df[wrong_hba1c_change]
````

**Observation :** It can be seen above that **56** values of `hba1c_change` are miscalculated. So, lets document it.

### 🎯 **Data Quality Dimensions**

*DataCamp Cheatsheet* on Data Quality Dimensions : [Link](https://media.datacamp.com/legacy/image/upload/v1678808802/Data_Quality_Dimensions_yebbuu.pdf)

There are mainly six types of data quality dimensions that are used to assess the quality of the data.  
1. Completeness
2. Uniqueness
3. Validity
4. Accuracy
5. Timeliness
6. Consistency

#### **1. Completeness**  
Completeness refers to the extent to which all required data is present, available, and recorded in the dataset.  

**Example :**  
| Order ID | Product | Price | Delivery Date |
| -------- | ------- | ----- | ------------- |
| 8001     | Laptop  | 60000 | 2024-06-10    |
| 8002     | Phone   | 32000 | *Missing*     |


#### **2. Uniqueness**  
Uniqueness measures the degree to which the records in a dataset are not duplicated.  

**Example :**  

| Customer ID | Name | Email                                 |
| ----------- | ---- | ------------------------------------- |

#### **4. Accuracy**  
Accuracy measures the degree to which data is correct and represents the truth.  
In other words, it refers to how correctly the data represents real world values.  

**Note :**  
1. A value can be *complete, valid, and consistent* yet still be ***inaccurate***.
2. Valid data can still be inaccurate, but invalid data is always inaccurate.  

**Example :** The actual price of a phone is ***1000 USD***. In data, it is stored as ***100 USD***. This is the problem of accuracy.  
Here, it can be seen that ***100 USD*** is *complete, valid, and consistent*, yet it is ***inaccurate*** because it doesn't matches with the actual price of the phone.


#### **5. Timeliness**

3. Even if the numbers are correct, they arrive too late to be useful.

#### **6. Consistency**

The `contact` column contains both *phone number* & *email*.   
So, we will be splitting the `contact` column into two different columns `phone` and `email`.  
For that, we will have to ***regex*** to extract phone number and email from the contact column.

````python
import re

# Phone Pattern
PHONE_RE = re.compile(r"""
    (?x)(?:\+?1[\s\-.]*)?(?:\(?\d{3}\)?[\s\-.]*)\d{3}[\s\-.]*\d{4}|(?<!\d)\d{10}(?!\d)
""")

# Email Pattern
EMAIL_RE = re.compile(r"[A-Za-z][A-Za-z0-9._%+\-]*@[A-Za-z0-9.\-]+\.[A-Za-z]{2,}")
````

````python
def _clean_markdown_mailto(s: str) -> str:
    """Turns [text](mailto:someone@x.com) into text; keeps parentheses used in phone numbers."""
    if s is None:
        return None
    s = str(s)
    s = re.sub(r"\[([^\]]+)\]\(mailto:[^)]+\)", r"\1", s)
    s = s.replace("mailto:", "")
    return s

def _normalize_us_phone(phone_raw: str) -> str:
    if not phone_raw:
        return None
    digits = re.sub(r"\D", "", phone_raw)
    if len(digits) == 11 and digits.startswith("1"):
        digits = digits[1:]
    if len(digits) != 10:
        return phone_raw.strip()
    return f"{digits[0:3]}-{digits[3:6]}-{digits[6:10]}"

def extract_email_phone(contact: str):
    s = _clean_markdown_mailto(contact)
    if s is None or s.strip() == "":
        return pd.Series({"email": None, "phone": None})

    email_match = EMAIL_RE.search(s)
    email = email_match.group(0) if email_match else None

    s_wo_email = s.replace(email, " ") if email else s
    phone_match = PHONE_RE.search(s_wo_email)
    phone_raw = phone_match.group(0) if phone_match else None
    phone = _normalize_us_phone(phone_raw)

    return pd.Series({"phone": phone, "email": email})
````

Starting with the `zip_code` column.  


While assessing the `zip_code` column, I found that the 4-digit zip codes are actually result of a behaviour where zeros in the starting are ignored as the data type of the `zip_code` column is `float`. For example:  
1. 01234 is interpreted as 1234 while reading the CSV file.
2. Similarly, 04765 is interpreted as 4765.  

Now, lets tackle this issue.  

**NOTE :** The point to be noted is that some of the missing values in the `zip_code` column is imputed as `0` when the data type was `float64`. But now, I will be changing the data type to `str`, so representation of missing values will be changed to `No Data`.

````python
def validate_zip_code(code):
    if code == '0':
        return 'No Data'
    elif len(code) < 5:
        return code.zfill(5)
    else:
        return code
````

To fix the issues with `auralin` and `novodra` column, the first approach that comes in our mind is to split `auralin` into `auralin_start` and `auralin_end`, and `novodra` into `novodra_start` and `novodra_end`, which is fine. But it will still contain `null` values, which isn't optimal.  

So, the optimal approach to deal with this issue is that, we should determine the medication type (either `auralin` or `novodra`). Then, we should have two columns `dosage_start` and `dosage_end` to store the start and end dosage values for the particular medication type.  

So, lets do this.

````python
def determine_dosage(row: pd.Series):
    '''Determines Medication Type and Dosage Values.'''
    
    if(row['auralin'] != '-'):
        row['medication'] = 'auralin'
    else:
        row['medication'] = 'novodra'
        
    start = float(row[row['medication']].split('-')[0].replace('u', '').strip())
    end = float(row[row['medication']].split('-')[1].replace('u', '').strip())
    
    row['dosage_start'] = start
    row['dosage_end'] = end
    
    return row
````
